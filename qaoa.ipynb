```python
import json
import phoenix
from qiskit import QuantumCircuit
```

```python
with open('./backup/qaoa_json/qaoa_reg3_16.json', 'r') as f:
    data = json.load(f)

ham = phoenix.Hamiltonian(data['paulis'], data['coeffs'])
qc = ham.generate_circuit()
phoenix.utils.print_circ_info(qc)
```

```python
qc.draw(fold=-1)
```

```python
qc_qiskit = phoenix.utils.compile_by_qiskit(data['paulis'], data['coeffs'])
phoenix.utils.print_circ_info(qc_qiskit)
qc_qiskit.draw(fold=-1)
```

```python
qc_qiskit.count_ops()
```

```python
qc_phoenix = phoenix.compile_hamiltonian_simulation(ham, grouping='support', optimize=False)
phoenix.utils.print_circ_info(qc_phoenix)
print(qc_phoenix.count_ops())
qc_phoenix.draw(fold=-1)
```

```python
qc_phoenix = phoenix.compile_hamiltonian_simulation(ham, optimize=False)
phoenix.utils.print_circ_info(qc_phoenix)
print(qc_phoenix.count_ops())
qc_phoenix.draw(fold=-1)
```
