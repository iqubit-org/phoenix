```python
from qiskit.circuit.library import RXGate, RYGate, RZGate, RXXGate, RYYGate, RZZGate
from qiskit.circuit.gate import Gate as QiskitGate
from qiskit.circuit import CircuitInstruction
from typing import Tuple

def pauli_rotation_gate_to_pauli_and_coeff(g: CircuitInstruction, num_qubits: int) -> Tuple[str, float]:
    assert isinstance(g.operation, (RXGate, RYGate, RZGate, RXXGate, RYYGate, RZZGate)), 'Not a Pauli rotation gate'
    pauli = ['I'] * num_qubits
    pauli_str = g.operation.name[1:].upper()
    for p, qb in zip(pauli_str, g.qubits):
        pauli[qb._index] = p
    return ''.join(pauli), g.operation.params[0] / 2
```

```python
from qiskit import QuantumCircuit
from qiskit.converters import circuit_to_dag

qc = QuantumCircuit(2, 1)
qc.h(0)
qc.cx(0, 1)
qc.measure(1, 0)

dag = circuit_to_dag(qc)

for node in dag.topological_op_nodes():
    print(type(node), isinstance(node.op, qiskit.circuit.Instruction), node.op.name)
```
