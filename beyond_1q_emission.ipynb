```python
import numpy as np
import phoenix
```

```python
paulis = [
    'XXII',
    'IXXY',
    'IYZX',
    'IYYX',
    'IZXZ'
]
phoenix.utils.plot_pauli_strings(paulis, figsize=(2.5,2.5), title=None, hide_axis=True, output_filename='original_paulis.svg')
```

```python
paulis = [
    'IIII',
    'IXXY',
    'IYZX',
    'IYYX',
    'IZXZ'
]
phoenix.utils.plot_pauli_strings(paulis, figsize=(2.5,2.5), title=None, hide_axis=True, output_filename='peeled_paulis.svg')
```

```python
ham = phoenix.Hamiltonian([p[::-1] for p in paulis], np.random.rand(len(paulis)))
ham.print_tableau()
```

```python
c1 = phoenix.CNOTEquivCliffordGate('z', 'x')
c2 = phoenix.CNOTEquivCliffordGate('y', 'x')
```

```python
ham_ = ham.apply_clifford(c1, 0, 1)
phoenix.utils.plot_pauli_strings([p[::-1] for p in ham_.paulis.to_labels()], figsize=(2.5,2.5), title=None, hide_axis=True, output_filename='update_c1_01.svg')
```

```python
ham_ = ham.apply_clifford(c1, 1, 0)
phoenix.utils.plot_pauli_strings([p[::-1] for p in ham_.paulis.to_labels()], figsize=(2.5,2.5), title=None, hide_axis=True, output_filename='update_c1_10.svg')
```

```python
ham_ = ham.apply_clifford(c2, 0, 1)
phoenix.utils.plot_pauli_strings([p[::-1] for p in ham_.paulis.to_labels()], figsize=(2.5,2.5), title=None, hide_axis=True, output_filename='update_c2_01.svg')
```

```python
ham_ = ham.apply_clifford(c2, 1, 0)
phoenix.utils.plot_pauli_strings([p[::-1] for p in ham_.paulis.to_labels()], figsize=(2.5,2.5), title=None, hide_axis=True, output_filename='update_c2_10.svg')
```
