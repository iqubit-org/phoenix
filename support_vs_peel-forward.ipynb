```python
import phoenix
from qiskit import QuantumCircuit
import numpy as np
```

```python
paulis = [
    "IZYZXX",
    "IYYZXY",
    # "IZYXZXZ",
    "ZZYXZI",
    "YYYXZI",
    # "YZYZXXI"
]
coeffs = np.random.rand(len(paulis)) / 100
ham = phoenix.Hamiltonian(paulis, coeffs)
```

```python
qc = phoenix.compile_hamiltonian_simulation(ham, optimize=False, grouping='support')
phoenix.utils.print_circ_info(qc)
phoenix.utils.print_circ_info(phoenix.optimize_phoenix_circuit_by_qiskit(qc))
qc.reverse_bits().draw(fold=-1)
```

```python
qc = phoenix.compile_hamiltonian_simulation(ham, optimize=False, grouping='peel', terminal='replay')
phoenix.utils.print_circ_info(qc)
phoenix.utils.print_circ_info(phoenix.optimize_phoenix_circuit_by_qiskit(qc))
qc.reverse_ops().reverse_bits().draw(fold=-1)
```

```python
from qiskit.transpiler.passes.synthesis import HLSConfig
from qiskit.transpiler import PassManager, passes

pm = PassManager([
    passes.CollectCliffords(matrix_based=True, min_block_size=1, max_block_width=qc.num_qubits),
    passes.HighLevelSynthesis(
        hls_config=HLSConfig(use_default_on_unspecified=False, clifford=["greedy"])
    ),
])
```

```python
pm.run(qc).draw(fold=-1)
```

```python
#   target: row 5 (YZYIZXXI), w=6
#   target: row 5 (YZYIZXXI), w=6
#   move 1: cxy@(1,2)  ΔW=-6 benefit=6 harm=0  weights → [5,5,5,5,5,5]
#   move 2: cxx@(2,3)  ΔW=-6 benefit=6 harm=0  weights → [4,4,4,4,4,4]
#   move 3: cyy@(3,5)  ΔW=-6 benefit=6 harm=0  weights → [3,3,3,3,3,3]
#   move 4: czx@(3,6)  ΔW=-6 benefit=6 harm=0  weights → [2,2,2,2,2,2]
#   t=4: 全部 6 行一个 cohort 发射(两个 pair 块,各 3 条旋转)
seq = ['cxy@(1,2)', 'cxx@(2,3)', 'cyy@(3,5)', 'czx@(3,6)']
h = ham
for op in seq:
    q0, q1 = map(int, op[5:-1].split(','))
    cliff = phoenix.CNOTEquivCliffordGate(*list(op[1:3]))
    h = h.apply_clifford(cliff, q0, q1)

h.generate_circuit().draw(fold=-1)
phoenix.optimize_phoenix_circuit_by_qiskit(h.generate_circuit()).draw(fold=-1)
```
