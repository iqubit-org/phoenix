```python
from qiskit.circuit import QuantumCircuit, QuantumRegister, ClassicalRegister

qr1 = QuantumRegister(2, "qreg1")
qr2 = QuantumRegister(1, "qreg2")
cr  = ClassicalRegister(3, "creg")
qc = QuantumCircuit(qr1, qr2, cr)

q = qc.qubits[2]          # 这是 qreg2 的第 0 个 qubit
loc = qc.find_bit(q)

print(loc.index)          # 全局索引（在 qc.qubits 里）
print(loc.registers)      # [(QuantumRegister(...), index_in_register)]
```

```python
qc.draw()
```
